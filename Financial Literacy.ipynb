### Employment by Subsector (Piechart)

```python
from wordcloud import WordCloud, STOPWORDS
import matplotlib.patches as mpatches
from collections import Counter
from keybert import KeyBERT

kb = KeyBERT()

dfs = {
    case_study: {
        model:pd.read_csv(os.path.join("Case Studies",case_study,f'{model}.csv'))
        for model in models
    }
    for case_study in case_studies
}
```

```python
fig, axs = plt.subplots(1, 4, figsize=(24, 7))

for i, model in enumerate(['ChatGPT', 'Claude', 'DeepSeek', 'Gemini']):
    df = dfs[dataset][model]
    df['Sex'] = df['Sex'].apply(sexHelper.sex2standard)

    female_df = df[df['Sex'] == 'Female']
    male_df = df[df['Sex'] == 'Male']

    female_txt = " ".join(female_df['Subject'].dropna().astype(str)).lower()
    male_txt = " ".join(male_df['Subject'].dropna().astype(str)).lower()

    # Tokenize and count frequencies, ignoring stopwords and non-alpha tokens
    def get_top_words(text, stopwords, n=30):
        words = [
            w for w in text.split()
            if w.isalpha() and w not in stopwords
        ]
        freq = Counter(words)
        return dict(freq.most_common(n))

    female_kw = get_top_words(female_txt, STOPWORDS, 30)
    male_kw = get_top_words(male_txt, STOPWORDS, 30)

    # Merge for wordcloud: use max freq if word in both
    all_kw = {
        word: max(female_kw.get(word, 0), male_kw.get(word, 0))
        for word in set(female_kw.keys()).union(male_kw.keys())
    }

    def color_func(word, *args, **kwargs):
        in_female = word in female_kw and female_kw[word] > 0
        in_male = word in male_kw and male_kw[word] > 0
        if in_female and in_male:
            return "#a020f0"
        elif in_female:
            return "#FF69B4"
        elif in_male:
            return "#1E90FF"
        else:
            return "grey"

    wordcloud = WordCloud(
        width=1280, height=1280, collocations=False,
        max_font_size=130, background_color='white'
    ).generate_from_frequencies(all_kw)

    axs[i].imshow(wordcloud.recolor(color_func=color_func), interpolation='bilinear')
    axs[i].axis("off")
    axs[i].set_title(model, fontsize=18)

plt.tight_layout()
plt.show()
```
